# 01. Exploratory Data Analysis (EDA) & Data Cleaning

In this notebook, we perform comprehensive data analysis, missing value handling, outlier detection, and duplicate removal on the Insurance Dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load raw dataset
df = pd.read_csv('../data/insurance.csv')
print('Dataset Shape:', df.shape)
df.head()
```

## Data Cleaning & Missing Value Check

```python
print('Missing Values:\n', df.isnull().sum())
print('Duplicate Count:', df.duplicated().sum())

# Clean duplicates
df_clean = df.drop_duplicates()
print('Cleaned Shape:', df_clean.shape)
```

## Statistical Distribution & Outlier Visualizations

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
sns.histplot(df_clean['charges'], kde=True, ax=axes[0, 0], color='purple').set_title('Charges Distribution')
sns.histplot(df_clean['bmi'], kde=True, ax=axes[0, 1], color='teal').set_title('BMI Distribution')
sns.boxplot(x='smoker', y='charges', data=df_clean, ax=axes[1, 0]).set_title('Smoker vs Charges Outlier Boxplot')
sns.scatterplot(x='age', y='charges', hue='smoker', data=df_clean, ax=axes[1, 1]).set_title('Age vs Charges by Smoker')
plt.tight_layout()
plt.show()
```
