# 05. Model Leaderboard Comparison & Joblib Model Persistence

In this notebook, we load saved leaderboard CSV files, visualize top champion models, and demonstrate loading fitted Scikit-Learn pipelines from Joblib for production inference.

```python
import pandas as pd
import joblib

# Load champion pipelines
reg_pipeline = joblib.load('../models/best_regression_pipeline.joblib')
cls_pipeline = joblib.load('../models/best_classification_pipeline.joblib')

# Perform inference on new patient sample
sample = pd.DataFrame([{
    'age': 45,
    'sex': 'male',
    'bmi': 31.2,
    'children': 3,
    'smoker': 'yes',
    'region': 'southeast'
}])

charge_pred = reg_pipeline.predict(sample)[0]
risk_pred = cls_pipeline.predict(sample)[0]

print(f'Predicted Insurance Charge: ${charge_pred:,.2f}')
print(f'Predicted Risk Classification: {"HIGH RISK" if risk_pred == 1 else "LOW RISK"}')
```
